```python
import pandas as pd
import numpy as np
```

```python
A_SEQUENCES = np.array([['B','B','B'],
                        ['B','B','R'],
                        ['B','R','B'],
                        ['B','R','R'],
                        ['R','B','B'],
                        ['R','B','R'],
                        ['R','R','B'],
                            ['R','R','R']])
B_SEQUENCES = A_SEQUENCES.copy()
```

```python
a_only = pd.DataFrame(data = np.zeros(shape=(8,4)),index = A_SEQUENCES, columns = ['Wins by Tricks','Ties by Tricks', 'Wins by Cards','Ties by Cards'])
a_only
```

```python
a = ['B','B','B']
```

```python
a_only.at[tuple(a), 'Wins by Tricks'] += 3
```

```python
match_sequences = []
for a in A_SEQUENCES:
        for b in B_SEQUENCES:
            if np.array_equal(a,b):
                continue
            match_sequences.append((tuple(a),tuple(b)))

df = pd.MultiIndex.from_tuples(match_sequences, names=['A','B'])
results = pd.DataFrame(0,index=df, columns=['Wins by Tricks','Wins by Cards','Ties by Tricks','Ties by Cards'])
results
```
